```python
def customer_segment(revenue):
    if revenue <= 304.25:
        return "Low Value"
    elif revenue <= 1631.475:
        return "Medium Value"
    else:
        return "High Value"

customer_segments = customer_revenue.apply(customer_segment)
```

```python
customer_segments.value_counts()
```

```python
customer_analysis = customer_revenue.reset_index()

customer_analysis["CustomerSegment"] = customer_analysis["Revenue"].apply(
    customer_segment
)

customer_analysis.head()
```

```python
segment_revenue = (
    customer_analysis
    .groupby("CustomerSegment")["Revenue"]
    .agg(["count", "sum", "mean", "median"])
    .sort_values("sum", ascending=False)
)

segment_revenue
```
